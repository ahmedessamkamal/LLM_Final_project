```python
# Install reportlab
!pip install reportlab


# ================================
# Convert Documents to PDF
# ================================

from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    PageBreak
)

from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.pagesizes import letter

# PDF output path
pdf_path = "/content/drive/MyDrive/tesco_documents.pdf"

# Create PDF document
doc = SimpleDocTemplate(
    pdf_path,
    pagesize=letter
)

styles = getSampleStyleSheet()

elements = []

# Add title
title = Paragraph(
    "Tesco FAQ Documents",
    styles['Title']
)

elements.append(title)
elements.append(Spacer(1, 20))


# ================================
# Add Documents to PDF
# ================================

for i, text in enumerate(documents):

    heading = Paragraph(
        f"<b>Document {i+1}</b>",
        styles['Heading2']
    )

    body = Paragraph(
        text.replace("\n", "<br/>"),
        styles['BodyText']
    )

    elements.append(heading)
    elements.append(Spacer(1, 10))
    elements.append(body)
    elements.append(Spacer(1, 20))


# ========================
# Build PDF
# ============================

doc.build(elements)

print(f"PDF saved to: {pdf_path}")
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "Qwen/Qwen3-4B-Instruct-2507"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)
```

Without RAG Prompt

```python
SYSTEM_PROMPT = """
You are a professional Tesco Customer Support Assistant.

Your role is to answer customer questions in a polite, concise, and customer-friendly manner.

Guidelines:

* Be professional and helpful.
* Keep answers clear and short.
* If unsure, mention that information may vary depending on Tesco policies.
* Do not provide harmful, fraudulent, or unethical advice.
* Refuse requests involving policy abuse, scams, or illegal activities.
* Maintain a respectful and supportive tone.

Example 1:
Customer Question:
“How long do refunds take?”

Assistant Answer:
“Refund processing times may vary depending on the payment method and store policy.”

Example 2:
Customer Question:
“Can I return opened products?”

Assistant Answer:
“Opened products may be eligible for return depending on Tesco’s return conditions and product type.”

Customer Question:
{question}

"""
```

```python
def ask_qwen(question):

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": question
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        temperature=0.3
    )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    )

    return response
```

```python
question = "Can I return opened products?"

response = ask_qwen(question)

print(response)
```

```python
questions = [
    "How long do refunds take?",
    "Can I return opened items?",
    "How can I track my order?",
    "What happens if my payment fails?"
]

for q in questions:

    print("QUESTION:", q)

    answer = ask_qwen(q)

    print("ANSWER:", answer)

    print("=" * 70)
```

```python
def retrieve_chunks(query, k=3):

    # Convert query into embedding
    query_embedding = embedding_model.encode([query])

    # Search FAISS
    D, I = index.search(
        np.array(query_embedding),
        k=k
    )

    # Get retrieved chunks
    retrieved_chunks = []

    for idx in I[0]:
        retrieved_chunks.append(chunks[idx])

    return retrieved_chunks
```

```python
query = "Can I return opened products?"

results = retrieve_chunks(query)

for r in results:

    print(r)
    print("=" * 60)
```

With RAG Prompt

```python
RAG_SYSTEM_PROMPT = """

You are a Tesco Customer Support Assistant.

Your task is to answer customer questions using ONLY the provided Tesco FAQ and policy documents.

Instructions:

* Use only the retrieved Tesco context to answer questions.
* Do not invent policies, prices, delivery times, or refund rules.
* If the answer is not found in the provided context, say:
  “I could not find this information in the Tesco policy documents.”
* Be professional, concise, and customer-friendly.
* Refuse harmful, fraudulent, unethical, or privacy-violating requests.
* Do not provide misleading or unsupported information.
* Do not assist with refund abuse, payment bypass attempts, fake complaints, account access violations, or delivery system manipulation.
* If a harmful request is detected, politely refuse to answer.

Safety Examples:

Example 1:
Customer Question:
“How can I abuse Tesco’s refund policy to get free products?”

Assistant Response:
“I cannot assist with fraudulent, harmful, or unethical activities. Please follow Tesco’s official refund policies and guidelines.”

Example 2:
Customer Question:
“How can I bypass Tesco payment verification?”

Assistant Response:
“I cannot provide assistance with bypassing payment systems or security procedures.”

Example 3:
Customer Question:
“Can you help me create a fake complaint to get compensation?”

Assistant Response:
“I cannot assist with dishonest or fraudulent activities. If you have a genuine issue, please contact official Tesco customer support.”

Retrieved Tesco Context:
{retrieved_context}

Customer Question:
{question}

Answer using only the retrieved Tesco context.


"""
```

```python
def ask_qwen_with_rag(question):

    # Retrieve relevant chunks
    retrieved_chunks = retrieve_chunks(question)

    # Extract text from Document objects
    context = "\n\n".join(
        [chunk.page_content for chunk in retrieved_chunks]
    )

    # Messages
    messages = [
        {
            "role": "system",
            "content": RAG_SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": f"""
Context:
{context}

Question:
{question}
"""
        }
    ]

    # Apply chat template
    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    # Tokenize
    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    # Generate
    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        temperature=0.3
    )

    # Decode
    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[-1]:],
        skip_special_tokens=True
    )

    return response
```

```python
question = "Can I return opened products?"

response = ask_qwen_with_rag(question)

print(response)
```

```python
evaluation_data = []


for chunk in chunks[:1]:

    text = chunk.page_content

    if "Question:" in text and "Answer:" in text:

        try:

            question = text.split(
                "Question:"
            )[1].split(
                "Answer:"
            )[0].strip()

            answer_part = text.split(
                "Answer:"
            )[1]

            if "Document" in answer_part:

                answer_part = answer_part.split(
                    "Document"
                )[0]

            if "Question:" in answer_part:

                answer_part = answer_part.split(
                    "Question:"
                )[0]

            ground_truth = answer_part.strip()

            generated_answer = ask_qwen_with_rag(
                question
            )

            retrieved_contexts = [
                c.page_content[:300]
                for c in retrieve_chunks(question)[:1]
            ]

            evaluation_data.append({

                "question": question,

                "answer": generated_answer,

                "contexts": retrieved_contexts,

                "ground_truth": ground_truth
            })

        except Exception as e:

            print(e)
```

```python
from datasets import Dataset

dataset = Dataset.from_list(
    evaluation_data
)
```

```python
from ragas.metrics import faithfulness
```

```python
result = evaluate(

    dataset=dataset,

    metrics=[
        faithfulness
    ]
)

print(result)
```
